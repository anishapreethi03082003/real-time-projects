```python
def is_prime(number):
    if number <= 1:
        return False
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            return False
    return True
```

```python
# Test the function
num = int(input("Enter a number: "))
```

```python
if is_prime(num):
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")
```

```python
import random

def multiplication_game():
    # Generate two random numbers between 1 and 10
    num1 = random.randint(1, 10)
    num2 = random.randint(1, 10)

    # Ask user for the product
    print(f"What is {num1} * {num2}?")
    user_answer = input("Enter your answer: ")

    # Check if the user's answer is correct
    try:
        if int(user_answer) == num1 * num2:
            print("Correct!")
        else:
            print(f"Sorry, that's incorrect. The correct answer is {num1 * num2}.")
    except ValueError:
        print("Invalid input. Please enter a number.")

    # Ask user if they want to play again
    play_again = input("Would you like to play again? (yes/no): ")
    if play_again.lower() == "yes":
        multiplication_game()
    else:
        print("Thanks for playing!")

# Start the game
multiplication_game()
```

```python
# Choose even numbers (change to "odd" if desired)
number_type = "even"

# Define the range
start = 100
end = 201 # Exclusive

# Loop through numbers in the range
for num in range(start, end):
    # Check if the number is even (or odd)
    if (num % 2 == 0 and number_type == "even") or (num % 2 != 0 and number_type == "odd"):
        # Print the square of the number
        print(f"The square of {num} is {num ** 2}")
```

```python
def count_words(text):
    # Split the text into words using spaces and count the number of words
    words = text.split()
    return len(words)

# Get input from the user
input_text = input(" This is a sample text. This text will be used to demonstrate the word counter.")

# Count the words in the input text
word_count = count_words(input_text)

# Display the result
print(f"The number of words in the text is: {word_count}")
```

```python
import string

def is_palindrome(s):
    # Convert to lowercase and remove punctuation and spaces
    s = s.lower()
    s = ''.join(char for char in s if char.isalnum())
    
    # Check if the string is equal to its reverse
    return s == s[::-1]

# Test the function
input_string = input("Enter a string: ")

if is_palindrome(input_string):
    print(f'"{input_string}" is a palindrome.')
else:
    print(f'"{input_string}" is not a palindrome.')
```
