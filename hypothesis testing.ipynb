```python
#Chi-Square test
```

```python
import numpy as np
import scipy.stats as stats
```

```python
# Step 1: Create the observed frequency table
observed = np.array([[50, 80, 60, 30, 20],
                     [70, 100, 90, 50, 50]])
```

```python
# Step 2: Perform the chi-square test for independence
chi2_stat, p_value, dof, expected = stats.chi2_contingency(observed)
```

```python
# Step 3: Determine the critical value
alpha = 0.05
critical_value = stats.chi2.ppf(1 - alpha, dof)
```

```python
# Step 4: Print the results
print(f"Chi-Square Statistic: {chi2_stat}")
print(f"Degrees of Freedom: {dof}")
print(f"Critical Value: {critical_value}")
print(f"P-Value: {p_value}")
print(f"Expected Frequencies:\n{expected}")
```

```python
# Step 5: Make a decision
if chi2_stat > critical_value:
    print("Reject the null hypothesis (H0). There is a significant association between the type of smart home device purchased and customer satisfaction.")
else:
    print("Fail to reject the null hypothesis (H0). There is no significant association between the type of smart home device purchased and customer satisfaction.")
```

```python
#Hypothesis Testing
```

```python
import numpy as np
import scipy.stats as stats
```

```python
# Given data
sample_mean = 3050 # Sample mean weekly cost
theoretical_mean = 1000 + 5 * 600 # Theoretical mean weekly cost (4000)
n = 25 # Sample size
std_dev_units = 25 # Standard deviation of the units produced
```

```python
# Step 1: Calculate the standard error of the mean
std_error = std_dev_units / np.sqrt(n)
```

```python
# Step 2: Calculate the test statistic (z-score)
z_score = (sample_mean - theoretical_mean) / std_error
```

```python
# Step 3: Determine the critical value for a one-tailed test at alpha = 0.05
alpha = 0.05
critical_value = stats.norm.ppf(1 - alpha)
```

```python
# Step 4: Print the results
print(f"Sample Mean: {sample_mean}")
print(f"Theoretical Mean: {theoretical_mean}")
print(f"Z-Score: {z_score}")
print(f"Critical Value (one-tailed, alpha=0.05): {critical_value}")
```

```python
# Step 5: Make a decision
if z_score > critical_value:
    print("Reject the null hypothesis (H0). There is significant evidence to support the claim that the weekly operating costs have increased.")
else:
    print("Fail to reject the null hypothesis (H0). There is no significant evidence to support the claim that the weekly operating costs have increased.")
```
