```python
# To plot the elbow method graph, we need to compute the WCSS (Within Cluster Sum of Squares)
# Let us say max. no. of clusters could be 10.
# As we are going to have 10 iterations we are going to write a for loop to create a list of 10 WCSS for the no. of clusters
from sklearn.cluster import KMeans# within cluster sum of square. Initialize WCSS and begin the loop
wcss=[]
for i in range(1,11):
  kmeans=KMeans(n_clusters=i,init='k-means++',random_state=42)
  kmeans.fit(scaled_data)
  wcss.append(kmeans.inertia_)
  #kmeans inertia initiate cluster centroid
   # In WCSS list we will append all distances i.e if cluster is 10 what is the value,
    # if 9 what is value and so on upto 1.

plt.plot(range(1, 11), wcss)
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
!pip install scikit-learn matplotlib


from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score, davies_bouldin_score # Import metrics



optimal_clusters = 3# we got 2 optimal number of clusters

# 2. Perform KMeans clustering
kmeans = KMeans(n_clusters=optimal_clusters, init='k-means++', random_state=42)
y_kmeans = kmeans.fit_predict(scaled_data) # Fit KMeans and get cluster labels

# Calculate Silhouette Score and Davies-Bouldin Index
silhouette_avg = silhouette_score(scaled_data, y_kmeans)
davies_bouldin_avg = davies_bouldin_score(scaled_data, y_kmeans)

print(f"Silhouette Score: {silhouette_avg}")
print(f"Davies-Bouldin Index: {davies_bouldin_avg}")

# 3. Apply PCA for dimensionality reduction
pca = PCA(n_components=2)  # Reduce to 2 dimensions for plotting
principal_components = pca.fit_transform(scaled_data)

# 4. Create a DataFrame with principal components and cluster labels
pca_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])
pca_df['Cluster'] = y_kmeans

# 5. Plot the clusters
plt.figure(figsize=(8, 6))
for cluster in range(optimal_clusters):
    cluster_data = pca_df[pca_df['Cluster'] == cluster]
    plt.scatter(cluster_data['PC1'], cluster_data['PC2'], label=f'Cluster {cluster + 1}')

plt.title('KMeans Clusters Visualization with PCA')
plt.xlabel('Principal Component 1 (PC1)')
plt.ylabel('Principal Component 2 (PC2)')
plt.legend()
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.cluster import AgglomerativeClustering
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage



# Perform hierarchical clustering
# linkage method: 'ward' (minimizes variance within clusters)
# affinity: 'euclidean' (distance metric)
linked = linkage(scaled_data, method='ward', metric='euclidean')

# Plot the dendrogram
plt.figure(figsize=(10, 7))
dendrogram(linked,
           orientation='top',
           distance_sort='descending',
           show_leaf_counts=True)
plt.title('Dendrogram')
plt.xlabel('Data Points')
plt.ylabel('Euclidean Distance')
plt.show()

# Create clusters based on the dendrogram (e.g., 2 clusters)
cluster = AgglomerativeClustering(n_clusters=3,linkage='ward')
clusters = cluster.fit_predict(scaled_data)

# Add the cluster labels to your DataFrame
scaled_data['cluster_labels'] = clusters

# Explore the clusters
print(scaled_data['cluster_labels'].value_counts())
# ... further analysis ...
```

**DBSCAN**

```python

from sklearn.decomposition import PCA
from sklearn.cluster import DBSCAN



#  Apply PCA for dimensionality reduction (if not already done)
pca = PCA(n_components=2)  # Reduce to 2 dimensions for plotting
principal_components = pca.fit_transform(scaled_data)

#  Apply DBSCAN clustering
dbscan = DBSCAN(eps=0.5, min_samples=5)  # Adjust eps and min_samples as needed
y_DBSCAN = dbscan.fit_predict(principal_components)

#  Create a DataFrame with principal components and cluster labels
pca_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])
pca_df['Cluster'] = y_DBSCAN

#  Plot the clusters
plt.figure(figsize=(8, 6))
unique_labels = set(y_DBSCAN )
colors = [plt.cm.Spectral(each)
          for each in np.linspace(0, 1, len(unique_labels))]
for k, col in zip(unique_labels, colors):
    if k == -1:
        # Black used for noise.
        col = [0, 0, 0, 1]

    class_member_mask = (y_DBSCAN  == k)

    xy = principal_components[class_member_mask]
    plt.plot(xy[:, 0], xy[:, 1], 'o', markerfacecolor=tuple(col),
             markeredgecolor='k', markersize=14)

plt.title('DBSCAN Clusters Visualization with PCA')
plt.xlabel('Principal Component 1 (PC1)')
plt.ylabel('Principal Component 2 (PC2)')
plt.show()
```

```python
# Calculate Silhouette Score and Davies-Bouldin Index
# Filter out noise points (cluster label -1) for these metrics
core_samples_mask = np.zeros_like(y_DBSCAN, dtype=bool)
core_samples_mask[dbscan.core_sample_indices_] = True
labels = y_DBSCAN[core_samples_mask]
data_for_metrics = principal_components[core_samples_mask]

silhouette_avg = silhouette_score(data_for_metrics, labels)
davies_bouldin_avg = davies_bouldin_score(data_for_metrics, labels)

print(f"Silhouette Score: {silhouette_avg}")
print(f"Davies-Bouldin Index: {davies_bouldin_avg}")

# ...
```

**Model-Based Clustering**

```python
from sklearn.decomposition import PCA
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score, davies_bouldin_score

# Apply PCA for dimensionality reduction
pca = PCA(n_components=2)
principal_components = pca.fit_transform(scaled_data)

# Apply Gaussian Mixture Model (GMM)
gmm = GaussianMixture(n_components=2, random_state=42)  # Adjust n_components as needed
y_gmm= gmm.fit_predict(principal_components)

# Calculate Silhouette Score and Davies-Bouldin Index
silhouette_avg = silhouette_score(principal_components, y_gmm )
davies_bouldin_avg = davies_bouldin_score(principal_components, y_gmm)

print(f"Silhouette Score: {silhouette_avg}")
print(f"Davies-Bouldin Index: {davies_bouldin_avg}")

# Create a DataFrame with principal components and cluster labels
pca_df = pd.DataFrame(data=principal_components, columns=['PC1', 'PC2'])
pca_df['Cluster'] = y_gmm

# Plot the clusters
plt.figure(figsize=(8, 6))
for cluster in range(3):  # Assuming 2 clusters
    cluster_data = pca_df[pca_df['Cluster'] == cluster]
    plt.scatter(cluster_data['PC1'], cluster_data['PC2'], label=f'Cluster {cluster + 1}')

plt.title('Gaussian Mixture Model Clusters Visualization with PCA')
plt.xlabel('Principal Component 1 (PC1)')
plt.ylabel('Principal Component 2 (PC2)')
plt.legend()
plt.show()
```
