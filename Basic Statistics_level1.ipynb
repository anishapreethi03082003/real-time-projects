```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('sales_data_with_discounts[1].csv')
```

```python
columns = ['Avg Price', 'Total Sale Value', 'Discount Rate', 'Discount Amount']
```

```python
def plot_histogram(data, column):
    plt.figure(figsize=(8, 6))
    sns.histplot(data[column], bins=30, kde=True) # kde=True adds a Kernel Density Estimate
    plt.title(f'Histogram of {column}')
    plt.xlabel(column)
    plt.ylabel('Frequency')
    plt.show()
```

```python
def plot_boxplot(data, column):
    plt.figure(figsize=(8, 6))
    sns.boxplot(x=data[column])
    plt.title(f'Boxplot of {column}')
    plt.xlabel(column)
    plt.show()
```

```python
plot_histogram(data,'Avg Price')
plot_boxplot(data,'Avg Price')
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv('sales_data_with_discounts[1].csv')
```

```python
pd.read_csv('sales_data_with_discounts[1].csv')
```

```python
df = pd.DataFrame(data)
```

```python
print("Original Data:")
print(df)
```

```python
numeric_columns = data.select_dtypes(include=['float64', 'int64']).columns
```

```python
scaler = StandardScaler()
```

```python
data[numeric_columns] = scaler.fit_transform(data[numeric_columns])
```

```python
print("Standardized dataset:")
print(data.head())
```

```python
# Standardizing manually: (X - mean) / std
for column in numeric_columns:
    data[column] = (data[column] - data[column].mean()) / data[column].std()
```

```python
print("Manually Standardized dataset:")
print(data.head())
```

```python
print("\nMean of each column after standardization:")
print(data[numeric_columns].mean())
```

```python
print("\nStandard deviation of each column after standardization:")
print(data[numeric_columns].std())
```
