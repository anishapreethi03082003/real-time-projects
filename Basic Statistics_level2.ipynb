```python
import numpy as np
import scipy.stats as stats
```

```python
#given data
data = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
```

```python
mean = np.mean(data)
print(f"Sample Mean: {mean:3f}")
```

```python
std_dev = np.std(data,ddof=1)
print(f"Sample Standard Deviation: {std_dev:.3f}")
```

```python
n = len(data)
print(f"Sample Size: {n}")
```

```python
#determine the t_value for 99% confidence interval with n-1 degrees of freedom
alpha = 0.01
df = n - 1
t_value = stats.t.ppf(1 - alpha / 2, df)
print(f"t-value for 99% CI and{df} degrees of freedom: {t_value:.3f}")
```

```python
margin_of_error = t_value*(std_dev / np.sqrt(n))
print(f"Margin of Error:{margin_of_error:.3f}")
```

```python
confidence_interval = (mean - margin_of_error , mean +margin_of_error)
print(f"99% Confidence Interval: {confidence_interval}")
```
